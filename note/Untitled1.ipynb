```python
import pandas as pd
from datetime import datetime, timedelta

# CSV 파일 로드
docker_df = pd.read_csv("docker_stats_6.csv")
test_df = pd.read_csv("6.csv")

# Timestamp를 한국시간(UTC+9)으로 변환
docker_df["Timestamp"] = pd.to_datetime(docker_df["Timestamp"], unit='s') + timedelta(hours=9)

test_df["DateTime"] = pd.to_datetime(test_df["DateTime"], format="%Y-%m-%d %H:%M:%S")
#test_df = test_df.set_index("DateTime").resample("1s").interpolate().reset_index()

# Timestamp와 DateTime을 가장 가까운 값으로 매칭
docker_df = docker_df.rename(columns={"Timestamp": "DateTime"})
merged_df = pd.merge_asof(
    test_df.sort_values("DateTime"),
    docker_df.sort_values("DateTime"),
    on="DateTime",
    direction="nearest",
    tolerance=pd.Timedelta("4s")  # 10초 이내 값만 매칭
)

# 결과 확인
print(merged_df)
```
