```python
import requests
res = requests.get("http://localhost:8949/nginx_status")
text = res.text
print(text)
```

```python
# ac
a_conn = text.split("\n")[0]
aa_conn = a_conn.split(": ")
ac = aa_conn[1].strip()

# ahr
b = text.split("\n")[1]
bb = b.split(" ")
ahr=bb[1:4]
c = text.split("\n")[2]
cc=c.strip()
ahr_2 = cc.split(" ")
ahr_result = dict(zip(ahr, ahr_2))
formatted_ahr = "\n".join(f"{key}={value}" for key, value in ahr_result.items())

#rww
rww=text.split("\n")[3]
rww_parts = rww.split()
rww_data = {rww_parts[i].strip(":"): rww_parts[i+1] for i in range(0, len(rww_parts), 2)}
formatted_rww = "\n".join(f"{key}={value}" for key, value in rww_data.items())
k = f'ac={ac}\n{formatted_ahr}\n{formatted_rww}'
print(k)
```
