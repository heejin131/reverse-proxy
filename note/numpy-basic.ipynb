### 정규분포

```python
import numpy as np
import matplotlib.pyplot as plt
```

```python
mean = 0 #평균
std_dev = 1 #표준편차
size = 10_000_000


data = np.random.normal(loc=mean, scale=std_dev, size = size)
```

```python
plt.hist(data, bins=100, edgecolor='green', alpha=0.7)
plt.title("N D")
plt.xlabel("V")
plt.ylabel("K")
plt.show()
```

```python
# 균등분포 - 0 ~ 10 사이에 균등분포를 갖는 1000개의 샘플
data = np.random.uniform(0, 10, 1000)
plt.hist(data, bins=30, edgecolor='black')
plt.xlim(0, 10)
plt.ylabel("Frequency")
plt.xlabel("Value")
plt.show()
```

```python
# 포아송분포
data = np.random.poisson(3,1000)
plt.hist(data, bins=30, edgecolor='black')
plt.xlim(0, 10)
plt.ylabel("Frequency")
plt.xlabel("Value")
plt.show()
```

```python
np.random.randint(0,10,10)
plt.hist(data, bins=30, edgecolor='black', alpha=0.7)
plt.show()
```

```python
import numpy as np
import random
import time
import matplotlib.pyplot as plt

# NumPy 기반 함수
def gen_r_array_choices_numpy(N):
    a = np.random.choice(range(101), size=N)
    b = np.random.choice(range(101), size=N)
    return a, b

# Python random 기반 함수
def gen_r_array_choices(N):
    a = random.choices(range(101), k=N)
    b = random.choices(range(101), k=N)
    return a, b

# randint
def gen_r_array_randint(N):
    a= np.random.randint(0, 101, size=N)
    b = np.random.randint(0, 101, size=N)
    return a, b

# 고차 함수: 실행 시간 측정
def measure_time(func, N_values):
    times = []
    for N in N_values:
        start = time.perf_counter()
        func(N)
        times.append(time.perf_counter() - start)
    return times

# 테스트할 N 값 생성 (1부터 1,000,000까지 100개 간격)
N_values = np.linspace(1, 1_000, 100, dtype=int)

# TODO 각 함수의 실행 시간 계산 위 배열에 추가
numpy_times = []
random_times = []
randint_times = []
numpy_times = measure_time(gen_r_array_choices_numpy, N_values)
random_times = measure_time(gen_r_array_choices, N_values)
randint_times = measure_time(gen_r_array_randint, N_values)

# 선그래프 그리기
plt.figure(figsize=(10, 6)) # 차트 크기 -> 그래프의 가로 크기 10인치, 세로 크기 6인치
plt.xlabel('Array Size') # x축 레이블
plt.ylabel('Time') # y축 레이블
plt.title('Performance : numpy VS random VS randint')
plt.grid(True) # 그래프에 그리드 ON
plt.plot(N_values, numpy_times, label='numpy', color='blue')
plt.plot(N_values, random_times, label='random', color='red')
plt.plot(N_values, randint_times, label='randint', color='Green')
plt.legend() # 범례를 추가
plt.show()
```
