```python
import pandas as pd
import matplotlib.pyplot as plt

# 🔹 CSV 파일 리스트
file_paths = ["docker_stats_6.csv", "docker_stats_7.csv"]
labels = ["Test 1", "Test 2"]  # 그래프 범례용 레이블
colors = ["blue", "red"]

# 🔹 그래프 데이터 저장 리스트
tps_data = []
mean_time_data = []
errors_data = []

# 🔹 1. CSV 파일 읽고 데이터 처리
for file in file_paths:
    df = pd.read_csv(file, delimiter="\t")  # nGrinder 데이터는 탭 구분 (TSV)

    # DateTime 변환 및 정렬
    df["Timestamp"] = pd.to_datetime(df["Timestamp"])
    df = df.sort_values("Timestamp")

    # 그래프용 데이터 저장
    tps_data.append((df["Timestamp"], df["TPS"]))
    mean_time_data.append((df["Timestamp"], df["Mean_Test_Time_(ms)"]))
    errors_data.append((df["Timestamp"], df["Errors"]))

# 🔹 2. 그래프 생성
plt.figure(figsize=(12, 6))  # 그래프 크기 설정

# 🔸 TPS 그래프
plt.subplot(3, 1, 1)
for i, (x, y) in enumerate(tps_data):
    plt.plot(x, y, label=labels[i], color=colors[i])
plt.ylabel("TPS")
plt.title("Performance Comparison: TPS, Mean Test Time, Errors")
plt.legend()
plt.grid(True)

# 🔸 Mean Test Time 그래프
plt.subplot(3, 1, 2)
for i, (x, y) in enumerate(mean_time_data):
    plt.plot(x, y, label=labels[i], color=colors[i])
plt.ylabel("Mean Test Time (ms)")
plt.legend()
plt.grid(True)

# 🔸 Errors 그래프
plt.subplot(3, 1, 3)
for i, (x, y) in enumerate(errors_data):
    plt.plot(x, y, label=labels[i], color=colors[i])
plt.xlabel("DateTime")
plt.ylabel("Errors")
plt.legend()
plt.grid(True)

# 🔹 3. 그래프 출력
plt.tight_layout()  # 서브플롯 간격 조정
plt.show()
```
