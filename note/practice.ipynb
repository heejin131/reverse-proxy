```python
import numpy as np
import random
import time
import matplotlib.pyplot as plt

# NumPy 기반 함수
def gen_r_array_choices_numpy(N):
    a = np.random.choice(range(101), size=N)
    b = np.random.choice(range(101), size=N)
    return a, b
```

```python
N = 10
gen_r_array_choices_numpy(N) 
# 1에서 101사이에서 숫자 10개를 랜덤하게 2개의 리스트를 만듦
```

```python
# 고차함수 함수안에 함수넣기

def measure_time(func, N_values):
    times = []
    for N in N_values:
        start = time.time() #시작 시간
        func(N)
        times.append(time.time() - start) # 현시간에서 시작시간을 뺌 결국, 측정후 시간 - 측정 전 시간 이므로 측정시간인 것
    return times
```

```python
def measure_time(func, N_values):
    times = []
    for N in N_values:
        start = time.perf_counter()
        func(N)
        times.append(time.perf_counter() - start)
    return times
```

```python
N_values = np.linspace(1, 10_000, 100, dtype=int)
t_times = measure_time(gen_r_array_choices_numpy, N_values)
t_times
```
