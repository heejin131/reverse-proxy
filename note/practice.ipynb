```python
# 고차함수 함수안에 함수넣기

def measure_time(func, N_values):
    times = []
    for N in N_values:
        start = time.time() #시작 시간
        func(N)
        times.append(time.time() - start) # 현시간에서 시작시간을 뺌 결국, 측정후 시간 - 측정 전 시간 이므로 측정시간인 것
    return times
```

```python
def measure_time(func, N_values):
    times = []
    for N in N_values:
        start = time.perf_counter()
        func(N)
        times.append(time.perf_counter() - start)
    return times
```

```python
import pandas as pd
import matplotlib.pyplot as plt

# 파일 경로 리스트 (파일 이름을 실제 경로로 변경)
file_paths = ["file1.csv", "file2.csv", "file3.csv", "file4.csv"]
labels = ["Test 1", "Test 2", "Test 3", "Test 4"]  # 그래프 범례용 레이블

# 색상 리스트
colors = ["blue", "red", "green", "orange"]

# 그래프 설정
plt.figure(figsize=(12, 6))

# 각 파일을 읽어서 그래프에 추가
for i, file in enumerate(file_paths):
    df = pd.read_csv(file, delimiter="\t")  # TSV 파일이라면 delimiter="\t"
    
    # DateTime 변환 (시간 순 정렬을 위해)
    df["DateTime"] = pd.to_datetime(df["DateTime"])
    df = df.sort_values("DateTime")

    # 그래프 그리기 (예제: Mean_Test_Time_(ms) 비교)
    plt.plot(df["DateTime"], df["Mean_Test_Time_(ms)"], marker='o', linestyle='-', color=colors[i], label=f"{labels[i]}")

# 그래프 설정
plt.xlabel("DateTime")
plt.ylabel("Mean Test Time (ms)")
plt.title("Performance Test Comparison (Mean Test Time)")
plt.legend()
plt.grid(True)

# 그래프 출력
plt.xticks(rotation=45)  # X축 라벨 회전
plt.show()
```
